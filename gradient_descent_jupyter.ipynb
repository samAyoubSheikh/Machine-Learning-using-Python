```python
import numpy as np
import matplotlib.pyplot as plt

costs=[]
def gradient_descent(x, y):

    m = 0
    b = 0

    iterations = 500
    n = len(x)
    learning_rate = 0.08

    costs =[]

    for i in range(iterations):
        y_pred = m * x + b

        #cost
        cost = (1/n)*sum([val**2 for val in (y - y_pred)]) 
        costs.append(cost)
        
        D_m = -(2/n) * sum(x * (y - y_pred)) #derivative with respect to m
        D_b = -(2/n) * sum(y - y_pred) #derivative with respect to b

        #updating m 
        m = m - learning_rate * D_m

        #updating b
        b = b - learning_rate * D_b

        print(f"Iteration{i}: m = {m:.4f}, b ={b:.4f},Cost ={cost:.4f}")

        return m,b,costs
x = np.array([1,2,3,4,5])

y = np.array([5,7,9,11,13])
m,b,costs = gradient_descent(x,y)
```

```python


plt.plot(range(500),costs)
plt.show()
```
