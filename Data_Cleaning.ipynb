**DATA CLEANING FOR MACHINE LEARNING USING PYTHON**

```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv("House_Price_Data_123_Rows.csv")
```

```python
df
```

**COPY THE DATASET**

```python
#why i copied this dataset is because i dont want to change the actual dataset
data = df.copy()
data
```

**REMOVE THE DUPLICATE ROWS IN DATASET**

```python
 # this will tell us is there any duplicated row by true or false
data.duplicated()

# this will give the total number of duplicated rows , but in my dataset i dont have any duplicated row
data.duplicated().sum() 


#now agr koi duplicate row hoti usko remove aise karty 

data = data.drop_duplicates() # ye duplicate rows ko delete krta hai ayi samaj
```

**REMOVING IRRELEVENT COLUMNS**

```python
#for example dataset mai aise columns hoty hai jinko hamari research ke sath zyada kuch nahi hota , 
#agr koi kaam nahi hai fir udaa dena hai na columns unko
#jaise house id ko house price prediction ke sath kya karna hai kuch nahi, toh udaa do isko

data.shape # total number of rows and columns dega 

data.columns # gives every column name
```

```python

#remove house Id column

data = data.drop("House_ID",axis="columns")
```

```python
data.shape
#now i can clearly see that the house id column is removed....😂😂😂
```

**CHECKING MISSING VALUES**

```python
data.count() # gives number of entries in each column
```

```python
data.isnull().sum() # gives values if there is any null value present 

#Inorder to replace any value 
#data.replave("can be any value",np.nan)
```

**FIXING INCONSISTENT FORMATS**

```python
#Like small capital letters 
#in garage column there is not any standard formatting , like YES ,Yes ,No ,NO etc
#to fix
data["Garage"] = data["Garage"].str.title()
```

```python
data
```

**CHECKING IMPOSSIBLE VALUES**

```python
#NEGATIVE VALUES IN SIZE_SQFT

data.loc[data["Size_sqft"]<=0,"Size_sqft"] = np.nan
#agr sizesqft column mai koi value 0 ya ussy chota huva usko null krdo
data.loc[data["Age_Years"] < 0 , "Age_Years"] = np.nan

data.head()
```

**CHECKING INVALID LABELS**
 

```python
#This means if the target variable contains null data or invalid data we will drop this thats it
data = data.dropna(subset=["Price_USD"])
```

```python
data.to_csv("Updated_House_Dataset.csv",index=False)
```
