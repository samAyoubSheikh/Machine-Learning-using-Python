```python
import numpy as np
import pandas as pd
from matplotlib import pyplot as plt
from sklearn import linear_model
```

```python
%matplotlib inline

area =[2600,3000,3200,3600,4000]
price =[550000,565000,610000,680000,725000]

plt.ylabel("Price($)")
plt.xlabel("Area(sq ft)")
plt.scatter(area,price, color ='red',marker ='+')
```

```python
#now use linear regression for this model
area1 = np.array(area).reshape(-1,1) # for converting area in two dimensional form 

reg = linear_model.LinearRegression()


reg.fit(area1,price)

reg.predict([[3300]]) # i wrote in square brackets because , it expects input in two dimensional array 
```

```python
m =reg.coef_ # gives the coefficient M
```

```python
b =reg.intercept_ # gives the intercept b
```

```python
#now lets cross check the predicted value with formula

y = m * price + b

135.78767123 * 3300 +180616.43835616432
```

```python
#lets see here the regression line

plt.ylabel("Price($)")
plt.xlabel("Area(sq ft)")
plt.scatter(area,price, color ='red',marker ='+')
plt.plot(area,reg.predict(area1),color ="blue")

```
